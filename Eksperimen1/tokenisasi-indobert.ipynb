```python
from transformers import AutoTokenizer

MODEL_NAME = "indobenchmark/indobert-base-p1"
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
```

```python
import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/Alfa4026/Skripsi/main/dataset/hatespeech_preprocessed.csv")
df
```

```python
label_cols = [
    "HS",
    "Abusive",
    "HS_Individual",
    "HS_Group",
    "HS_Religion",
    "HS_Race",
    "HS_Physical",
    "HS_Gender",
    "HS_Other",
    "HS_Weak",
    "HS_Moderate",
    "HS_Strong"
]

X = df["stopword_clean"].values
Y = df[label_cols].astype(int).values
```

```python
texts = df["stopword_clean"].astype(str).tolist()
```

```python
MAX_LEN = 128

encodings = tokenizer(
    texts,
    truncation=True,
    padding="max_length",
    max_length=MAX_LEN,
    return_attention_mask=True,
)
```

```python
import pandas as pd

df_tokens = pd.DataFrame({
    "input_ids": encodings["input_ids"],
    "attention_mask": encodings["attention_mask"]
})

# Gabungkan label
df_final = pd.concat([df_tokens, df[label_cols].reset_index(drop=True)], axis=1)

df_final.head()
```
